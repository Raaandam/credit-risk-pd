# 01 - Exploratory data analysis: Lending Club accepted loans

Population checks before modelling: vintage volumes, default rates, missingness and leakage candidates. Set `USE_SYNTHETIC = True` to run without the Kaggle file.

```python
import sys
from pathlib import Path

import matplotlib.pyplot as plt
import pandas as pd

sys.path.insert(0, str(Path.cwd().parent / 'src'))
from credit_risk_pd import data
from credit_risk_pd.synthetic import make_synthetic_lending_club

USE_SYNTHETIC = False
RAW = Path('../data/raw/accepted_2007_to_2018Q4.csv')
```

## Load (sampled) data

```python
if USE_SYNTHETIC or not RAW.exists():
    raw = make_synthetic_lending_club(20_000)
else:
    raw = pd.read_csv(RAW, nrows=300_000, low_memory=False)
raw.shape
```

## Population: 36-month loans, 2013-2016, resolved outcomes

```python
df = data.prepare_dataset(raw)
df.groupby('issue_year')[data.TARGET].agg(['size', 'mean'])
```

## Default rate by grade and purpose

```python
fig, axes = plt.subplots(1, 2, figsize=(12, 4))
df.groupby('grade')[data.TARGET].mean().plot.bar(ax=axes[0], title='Default rate by grade')
df.groupby('purpose')[data.TARGET].mean().sort_values().plot.barh(ax=axes[1], title='Default rate by purpose')
plt.tight_layout()
```

## Missingness of allowlisted features

```python
num, cat = data.available_features(df)
df[num + cat].isna().mean().sort_values(ascending=False).head(15)
```
